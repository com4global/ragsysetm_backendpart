```python
# ============================================================================
# STEP 4: EMBEDDING GENERATION & VALIDATION
# ============================================================================

def generate_and_validate_embeddings(chunks: List[str]) -> Optional[List[List[float]]]:
    """Generate embeddings with comprehensive validation"""
    print(f"\n{'='*80}")
    print("STEP 4: EMBEDDING GENERATION & VALIDATION")
    print(f"{'='*80}\n")
    
    if not chunks:
        result = StepResult(
            step_name="Embedding Generation",
            success=False,
            message="No chunks to embed",
            error="Chunks list is empty"
        )
        validator.add_result(result)
        return None
    
    try:
        logger.info(f"Generating embeddings for {len(chunks)} chunks...")
        embeddings = embed_chunks(chunks)
        
        # Validation
        assert embeddings is not None, "Embeddings is None"
        assert len(embeddings) > 0, "Zero embeddings generated"
        assert len(embeddings) == len(chunks), "Embedding count mismatch"
        
        # Check dimensions
        embedding_dim = len(embeddings[0])
        expected_dim = CONFIG["embedding_dim"]
        assert embedding_dim == expected_dim, f"Dimension mismatch: {embedding_dim} vs {expected_dim}"
        
        # Validate all embeddings
        valid_embeddings = 0
        invalid_embeddings = []
        
        for i, emb in enumerate(embeddings):
            if emb is None:
                invalid_embeddings.append(f"Embedding {i}: None")
            elif len(emb) != embedding_dim:
                invalid_embeddings.append(f"Embedding {i}: wrong dimension {len(emb)}")
            else:
                valid_embeddings += 1
        
        # Statistical analysis
        first_values = [e[0] for e in embeddings if e is not None]
        
        metrics = {
            "total_embeddings": len(embeddings),
            "valid_embeddings": valid_embeddings,
            "embedding_dimensions": embedding_dim,
            "min_value": f"{min(first_values):.6f}",
            "max_value": f"{max(first_values):.6f}",
            "mean_value": f"{statistics.mean(first_values):.6f}",
            "std_deviation": f"{statistics.stdev(first_values):.6f}" if len(first_values) > 1 else "N/A",
        }
        
        print(f"✅ EMBEDDING GENERATION SUCCESSFUL")
        for key, value in metrics.items():
            print(f"   {key}: {value}")
        
        # Sample embedding
        print(f"\n🔢 Sample embedding (first 10 values):")
        print(f"   {[round(v, 6) for v in embeddings[0][:10]]}")
        
        # Warnings
        if invalid_embeddings:
            logger.warning(f"⚠️  Invalid embeddings found: {len(invalid_embeddings)}")
            for inv in invalid_embeddings[:3]:
                logger.warning(f"   {inv}")
        
        result = StepResult(
            step_name="Embedding Generation",
            success=len(invalid_embeddings) == 0,
            message=f"Generated {valid_embeddings} valid embeddings",
            data=embeddings,
            metrics=metrics
        )
        validator.add_result(result)
        
        return embeddings
        
    except AssertionError as e:
        error_msg = f"Assertion error: {str(e)}"
        logger.error(error_msg)
        result = StepResult(
            step_name="Embedding Generation",
            success=False,
            message="Embedding validation failed",
            error=error_msg
        )
        validator.add_result(result)
        return None
        
    except Exception as e:
        error_msg = f"Embedding generation error: {str(e)}"
        logger.error(error_msg)
        result = StepResult(
            step_name="Embedding Generation",
            success=False,
            message="Failed to generate embeddings",
            error=error_msg
        )
        validator.add_result(result)
        return None

# Execute embedding generation
if chunks:
    embeddings = generate_and_validate_embeddings(chunks)
else:
    embeddings = None
    logger.error("❌ Cannot embed: no chunks available")
```

```python
# ============================================================================
# STEP 3: TEXT CHUNKING & VALIDATION
# ============================================================================

def chunk_and_validate(pages: List[str]) -> Optional[List[str]]:
    """Chunk text with comprehensive validation"""
    print(f"\n{'='*80}")
    print("STEP 3: TEXT CHUNKING & VALIDATION")
    print(f"{'='*80}\n")
    
    if not pages:
        result = StepResult(
            step_name="Text Chunking",
            success=False,
            message="No pages to chunk",
            error="Pages list is empty"
        )
        validator.add_result(result)
        return None
    
    try:
        # Perform chunking
        chunk_size = CONFIG["chunk_size"]
        chunk_overlap = CONFIG["chunk_overlap"]
        
        logger.info(f"Chunking with size={chunk_size}, overlap={chunk_overlap}")
        chunks = chunk_pages(pages, chunk_size=chunk_size, chunk_overlap=chunk_overlap)
        
        # Validation
        assert chunks is not None, "Chunks is None"
        assert len(chunks) > 0, "Zero chunks created"
        
        # Analysis
        chunk_sizes = [len(c) for c in chunks]
        
        metrics = {
            "total_chunks": len(chunks),
            "min_size": f"{min(chunk_sizes):,}",
            "max_size": f"{max(chunk_sizes):,}",
            "avg_size": f"{statistics.mean(chunk_sizes):,.0f}",
            "median_size": f"{statistics.median(chunk_sizes):,.0f}",
            "total_chars": f"{sum(chunk_sizes):,}",
        }
        
        print(f"✅ CHUNKING SUCCESSFUL")
        for key, value in metrics.items():
            print(f"   {key}: {value}")
        
        # Chunk preview
        print(f"\n📦 Sample chunks:")
        for i, chunk in enumerate(chunks[:2]):
            print(f"\n   Chunk {i} ({len(chunk):,} chars):")
            print(f"   {chunk[:150]}...")
        
        # Validation checks
        issues = []
        if min(chunk_sizes) < 100:
            issues.append("Some chunks very small (<100 chars)")
        if max(chunk_sizes) > chunk_size * 2:
            issues.append(f"Some chunks exceed 2x chunk_size")
        
        if issues:
            logger.warning(f"⚠️  Potential issues: {', '.join(issues)}")
        
        result = StepResult(
            step_name="Text Chunking",
            success=True,
            message=f"Created {len(chunks)} chunks",
            data=chunks,
            metrics=metrics
        )
        validator.add_result(result)
        
        return chunks
        
    except Exception as e:
        error_msg = f"Chunking error: {str(e)}"
        logger.error(error_msg)
        result = StepResult(
            step_name="Text Chunking",
            success=False,
            message="Failed to chunk text",
            error=error_msg
        )
        validator.add_result(result)
        return None

# Execute chunking
if pages:
    chunks = chunk_and_validate(pages)
else:
    chunks = None
    logger.error("❌ Cannot chunk: no pages extracted")
```

```python
# ============================================================================
# SECTION 2: DIAGNOSTIC DATA STRUCTURES & UTILITIES
# ============================================================================

@dataclass
class StepResult:
    """Result of a pipeline step"""
    step_name: str
    success: bool
    message: str
    data: Optional[Any] = None
    error: Optional[str] = None
    metrics: Optional[Dict] = None

class RAGPipelineValidator:
    """Senior-level validator for RAG pipeline"""
    
    def __init__(self):
        self.results: List[StepResult] = []
        self.errors: List[str] = []
        
    def add_result(self, result: StepResult):
        self.results.append(result)
        if not result.success:
            self.errors.append(f"{result.step_name}: {result.error or result.message}")
    
    def print_report(self):
        """Print comprehensive validation report"""
        print(f"\n{'='*80}")
        print("RAG PIPELINE VALIDATION REPORT")
        print(f"{'='*80}")
        
        for result in self.results:
            status_symbol = "✅" if result.success else "❌"
            print(f"\n{status_symbol} {result.step_name}")
            print(f"   Message: {result.message}")
            
            if result.metrics:
                for key, value in result.metrics.items():
                    print(f"   {key}: {value}")
            
            if result.error:
                print(f"   Error: {result.error}")
        
        print(f"\n{'='*80}")
        print(f"Summary: {sum(1 for r in self.results if r.success)}/{len(self.results)} steps passed")
        if self.errors:
            print(f"\n⚠️ Issues found:")
            for error in self.errors:
                print(f"   - {error}")
        print(f"{'='*80}\n")

validator = RAGPipelineValidator()

logger.info("✅ Diagnostic tools initialized")
```
